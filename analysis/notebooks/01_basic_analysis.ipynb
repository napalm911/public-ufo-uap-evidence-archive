# UFO/UAP Evidence Analysis

Analyze downloaded U.S. government UAP documents using vector search, NLP, and timeline analysis.

**Setup:** First run `python scripts/extract_text.py` to extract text from all PDFs.

```python
# Load metadata
import json
from pathlib import Path

BASE_DIR = Path.cwd().parent
METADATA_DIR = BASE_DIR / "metadata"
EXTRACTED_DIR = BASE_DIR / "analysis" / "extracted_text"

with open(METADATA_DIR / "master_index.json") as f:
    master = json.load(f)
with open(METADATA_DIR / "timeline.json") as f:
    timeline = json.load(f)
with open(METADATA_DIR / "topic_tags.json") as f:
    tags = json.load(f)

print(f"Sources: {master['total_sources']}")
print(f"Files: {master['total_files']}")
print(f"Dated timeline entries: {timeline['total_dated_entries']}")
```

```python
# Analyze document distribution by year
from collections import Counter

years = []
for entry in timeline['entries']:
    year = entry['date'][:4]
    years.append(year)

year_counts = Counter(years)
for year in sorted(year_counts):
    count = year_counts[year]
    bar = "█" * count
    print(f"{year}: {bar} {count}")
```

```python
# Load extracted text into memory for analysis
texts = {}
txt_files = list(EXTRACTED_DIR.rglob("*.txt"))
print(f"Found {len(txt_files)} extracted text files")

for txt_file in txt_files[:10]:  # Sample first 10
    text = txt_file.read_text(encoding="utf-8")
    texts[txt_file.stem] = text
    print(f"  {txt_file.stem}: {len(text)} chars")
```

```python
# Example: find all mentions of 'non-human' or 'non-human intelligence'
import re

for name, text in texts.items():
    matches = re.findall(r'(?i)non-?human[^.]*', text)
    if matches:
        print(f"\n=== {name} ===")
        for m in matches[:5]:
            print(f"  • {m.strip()[:200]}")
```

2. **Language Evolution**: Track how government language evolved from "UFO" → "UAP" → "anomalous phenomena"
3. **Whistleblower Cross-Reference**: Compare claims of whistleblowers (Grusch, Elizondo, etc.) against official documents
4. **Nuclear Connection**: Analyze documents linking UAP activity to nuclear facilities
5. **Congressional Pressure**: Track the increase in legislative action on UAP over time
